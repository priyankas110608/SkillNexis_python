```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('data.csv')
print(df.info())
print(df.head())

# 2. Clean the Date column (convert strings to datetime, fix bad formats, drop unparseable rows)
df['Date'] = pd.to_datetime(df['Date'], format='mixed', errors='coerce')
df = df.dropna(subset=['Date'])

# 3. Handle missing values in numeric columns (e.g., fill Calories with median or mean)
df['Calories'] = df['Calories'].fillna(df['Calories'].median())
df = df.dropna()  # Drop any remaining rows with missing values

# 4. Fix obvious outliers/typos (e.g., if Duration has an extra zero like 450 instead of 45)
df.loc[df['Duration'] > 300, 'Duration'] = 45

# 5. Remove duplicate rows and add a useful derived column
df = df.drop_duplicates()
df['Pulse_Ratio'] = df['Pulse'] / df['Maxpulse']

print("\nCleaned data:")
print(df.info())
print(df.head())
```
